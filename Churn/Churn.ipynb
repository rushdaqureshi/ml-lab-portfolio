```python
## Load libraries
import pandas as pd
import numpy as np
```

```python
# Load the dataset
df = pd.read_csv('TCC.csv')

# Display first 5 rows
df.head()
```

```python
# Check shape (rows, columns)
print("Shape:", df.shape)

# Check data types and missing values
df.info()
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# Check for missing values
df.isna().sum()
```

```python
# Convert TotalCharges to numeric (bad values become NaN)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Check how many NaN values now exist
print("Missing TotalCharges:", df['TotalCharges'].isna().sum())

# Fill missing values with median
median_charges = df['TotalCharges'].median()

# FIX: Use direct assignment instead of inplace=True
df['TotalCharges'] = df['TotalCharges'].fillna(median_charges)

# Verify no missing values remain
print("After imputation:", df['TotalCharges'].isna().sum())
```

```python
# FIRST: Make all column names lowercase (makes life easier)
df.columns = df.columns.str.lower()

# SECOND: Now get categorical columns (all object types)
categorical_cols = df.select_dtypes(include=['object']).columns

# THIRD: Clean each categorical column
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()

# FOURTH: Verify by checking unique values
print("Payment method unique values:", df['paymentmethod'].unique())
print("Contract type unique values:", df['contract'].unique())
```

```python
# Count duplicate rows
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

# Remove duplicates if any
if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicates removed.")
else:
    print("No duplicates found.")
```

```python
# Define target column (use lowercase!)
target_col = 'churn'

# Define columns to drop (use lowercase!)
columns_to_drop = ['customerid', target_col]

# Create feature matrix X (all columns except dropped ones)
X = df.drop(columns=columns_to_drop)

# Create target vector y
y = df[target_col]

# Verify shapes
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nFeatures:", list(X.columns))
print("\nTarget value counts:")
print(y.value_counts())
```

```python
# Save the cleaned DataFrame (including target) for later use
output_path = 'TCC.csv'  # Saves in the same folder as your notebook
df.to_csv(output_path, index=False)
print(f"Cleaned data saved to {output_path}")
```
