```python
# ============================================
# LAB 01: Data Loading and Cleaning - Titanic Dataset (Fixed)
# ============================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 1: LOADING THE DATA")
print("=" * 60)

# Load the training data
df_train = pd.read_csv('train.csv')
print(f"Training data loaded: {df_train.shape[0]} rows, {df_train.shape[1]} columns")

# Load the test data
df_test = pd.read_csv('test.csv')
print(f"Test data loaded: {df_test.shape[0]} rows, {df_test.shape[1]} columns")

print("\n" + "=" * 60)
print("STEP 2: FIRST LOOK AT THE DATA")
print("=" * 60)

print("\n--- Training Data Info ---")
print(df_train.info())

print("\n--- Missing Values in Training Data ---")
missing_train = df_train.isna().sum()
print(missing_train[missing_train > 0])

print("\n" + "=" * 60)
print("STEP 3: HANDLING MISSING VALUES")
print("=" * 60)

# Make a copy to work with
df_train_clean = df_train.copy()

# 3.1: Fix Age - Fixed: Removed inplace=True and used assignment
age_median = df_train_clean['Age'].median()
df_train_clean['Age'] = df_train_clean['Age'].fillna(age_median)
print(f"✓ Filled missing Age values with median: {age_median}")

# 3.2: Fix Embarked - Fixed: Removed inplace=True and used assignment
embarked_mode = df_train_clean['Embarked'].mode()[0]
df_train_clean['Embarked'] = df_train_clean['Embarked'].fillna(embarked_mode)
print(f"✓ Filled missing Embarked values with mode: '{embarked_mode}'")

# 3.3: Cabin - Drop column (inplace is still fine for drop() on a full DataFrame)
df_train_clean.drop(columns=['Cabin'], inplace=True)
print("✓ Dropped Cabin column")

print("\n" + "=" * 60)
print("STEP 4: CHECKING FOR DUPLICATES")
print("=" * 60)

duplicates = df_train_clean.duplicated().sum()
if duplicates > 0:
    df_train_clean.drop_duplicates(inplace=True)
    print(f"✓ Removed {duplicates} duplicates")
else:
    print("No duplicates found.")

print("\n" + "=" * 60)
print("STEP 5: STANDARDIZING TEXT DATA")
print("=" * 60)

df_train_clean['Sex'] = df_train_clean['Sex'].str.lower()
print("✓ Converted Sex column to lowercase")

print("\n" + "=" * 60)
print("STEP 6: SEPARATING FEATURES (X) AND TARGET (y)")
print("=" * 60)

target_col = 'Survived'
identifiers = ['PassengerId', 'Name', 'Ticket']

X = df_train_clean.drop(columns=identifiers + [target_col])
y = df_train_clean[target_col]

print(f"Features (X) shape: {X.shape}")
print(f"Target (y) distribution:\n{y.value_counts()}")

print("\n" + "=" * 60)
print("STEP 7: SAVING THE CLEANED DATA")
print("=" * 60)

# Clean the test data using the SAME logic and assignment method
df_test_clean = df_test.copy()

# Fix Age and Fare in Test data using assignment
df_test_clean['Age'] = df_test_clean['Age'].fillna(age_median)

if df_test_clean['Fare'].isna().sum() > 0:
    fare_median = df_test_clean['Fare'].median()
    df_test_clean['Fare'] = df_test_clean['Fare'].fillna(fare_median)
    print(f"✓ Filled missing Fare values in Test set")

df_test_clean.drop(columns=['Cabin'], inplace=True)
df_test_clean['Sex'] = df_test_clean['Sex'].str.lower()

# Save files
df_train_clean.to_csv('training_cleaned.csv', index=False)
df_test_clean.to_csv('titanic_test_cleaned.csv', index=False)
print("✓ Files saved successfully!")
```
