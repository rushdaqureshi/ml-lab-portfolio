```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset
df = pd.read_csv('SP.csv')  # or use the full path if needed

# Display first 5 rows
df.head()
```

```python
# Check shape
print("Shape:", df.shape)

# Check data types and missing values
df.info()

# Check for missing values
print("\nMissing values:\n", df.isna().sum())

# Summary statistics for numerical columns
df.describe()
```

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

# Remove duplicates if any
if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicates removed.")
else:
    print("No duplicates found.")
```

```python
# Convert all column names to lowercase for consistency
df.columns = df.columns.str.lower()
print("Column names:", df.columns.tolist())
```

```python
# Get all categorical (object type) columns
categorical_cols = df.select_dtypes(include=['object']).columns
print("Categorical columns:", list(categorical_cols))

# Clean each one: strip spaces and convert to lowercase
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()

# Verify
print("\nUnique values in 'gender':", df['gender'].unique())
print("Unique values in 'lunch':", df['lunch'].unique())
print("Unique values in 'test preparation course':", df['test preparation course'].unique())
```

```python
# Choose one subject as target
target_col = 'math score'  # or 'reading score' or 'writing score'

# OR create an average score
df['avg_score'] = (df['math score'] + df['reading score'] + df['writing score']) / 3
target_col = 'avg_score'
```

```python
# Create average score first
df['avg_score'] = (df['math score'] + df['reading score'] + df['writing score']) / 3

# Create categories based on average score
def categorize_score(score):
    if score >= 80:
        return 'high'
    elif score >= 60:
        return 'medium'
    else:
        return 'low'

df['performance'] = df['avg_score'].apply(categorize_score)
target_col = 'performance'
```

```python
# Define target column (choose one based on Step 6)
target_col = 'performance'  # or 'math score' or 'avg_score'

# Define columns to drop (target + any non-predictive columns)
# For this dataset, all columns except target can be features
columns_to_drop = [target_col]

# If you created 'avg_score', drop it from features (keep only original scores)
if 'avg_score' in df.columns:
    columns_to_drop.append('avg_score')

# Create feature matrix X
X = df.drop(columns=columns_to_drop)

# Create target vector y
y = df[target_col]

# Verify shapes
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nFeatures:", list(X.columns))
print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Save the cleaned DataFrame
output_path = 'student_performance_cleaned.csv'
df.to_csv(output_path, index=False)
print(f"Cleaned data saved to {output_path}")
```
